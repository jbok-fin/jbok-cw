```python
from pyspark.sql import functions as F
from pyspark.sql import types as T

CATALOG = "sandbox"
SCHEMA = "sandbox_jbok"
RUN_FULL_COUNTS = True  # Set False to reduce expensive full-table scans.

MODEL_SPECS = {
    "host_node_features_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "node", "region", "zone"],
        "measures": [
            "context_switches_delta",
            "cpu_pressure_waiting_delta",
            "cpu_pressure_waiting_delta",
            "io_pressure_waiting_delta",
            "disk_read_bytes_delta",
            "disk_written_bytes_delta",
            "network_receive_bytes_delta",
            "network_transmit_bytes_delta",
        ],
    },
    "network_node_demand_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "node", "region", "zone"],
        "measures": [
            "total_in_bytes", "total_out_bytes", "total_bytes",
            "traffic_bps_in", "traffic_bps_out", "connections",
            "interface_count",
        ],
    },
    "network_cluster_context_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "region", "zone"],
        "measures": [
            "total_in_bytes_sum", "total_out_bytes_sum", "total_bytes_sum",
            "traffic_bps_in_sum", "traffic_bps_out_sum", "connections_sum",
            "activelogic_egress_ingress_ratio",
            "total_egress_ingress_ratio",
        ],
    },
    "network_cluster_org_demand_15m": {
        "grain": ["datestamp", "cluster_org", "region", "zone"],
        "measures": [
            "total_in_bytes_sum", "total_out_bytes_sum", "total_bytes_sum",
            "traffic_bps_in_sum", "traffic_bps_out_sum", "connections_sum",
        ],
    },
    "network_cluster_org_features_15m": {
        "grain": ["datestamp", "cluster_org", "region", "zone"],
        "measures": [
            "total_in_bytes_sum", "total_out_bytes_sum", "total_bytes_sum",
            "traffic_bps_in_sum", "traffic_bps_out_sum",
            "activelogic_egress_ingress_ratio",
            "total_egress_ingress_ratio",
        ],
    },
    "cpu_node_demand_signals_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "node", "region", "zone"],
        "measures": [
            "cpu_util_rate",
            "cpu_load1_per_cpu_ratio",
            "busy_cpu_seconds_total",
            "idle_cpu_seconds_total",
            "cpu_busy_ratio",
            "cpu_headroom_ratio",
        ],
    },
    "cpu_cluster_context_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "region", "zone"],
        "measures": [
            "cpu_util_rate",
            "cpu_load1_per_cpu_ratio",
            "cpu_busy_ratio",
            "cpu_headroom_ratio",
        ],
    },
    "gpu_hostnode_classifier_base_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "node", "region", "zone"],
        "measures": [
            "gpu_count", "gpu_util", "tensor_util", "chip_power",
            "redfish_power", "mem_copy_util", "sm_active", "dram_active",
            "is_training", "is_sunk_running",
            "is_hpc_verification_job", "is_multinode",
        ],
    },
    "cluster_workload_context_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "region", "zone"],
        "measures": [
            "cpu_util_rate",
            "cpu_busy_ratio",
            "traffic_bps_in_sum",
            "traffic_bps_out_sum",
            "connections_sum",
            "activelogic_egress_ingress_ratio",
        ],
    },
    "gpu_hostnode_classifier_features_15m": {
        "grain": ["datestamp", "cluster", "cluster_org", "node", "region", "zone"],
        "measures": [
            "gpu_util", "tensor_util", "chip_power",
            "context_switches_delta",
            "disk_written_bytes_delta",
            "network_receive_bytes_delta",
            "network_transmit_bytes_delta",
            "is_training", "is_sunk_running",
            "is_hpc_verification_job", "is_multinode",
            "traffic_bps_in_sum", "traffic_bps_out_sum",
            "activelogic_egress_ingress_ratio",
        ],
    },
}

def table_name(model):
    return f"`{CATALOG}`.`{SCHEMA}`.`{model}`"

def load_table(model):
    try:
        return spark.table(table_name(model)), None
    except Exception as exc:
        return None, str(exc).splitlines()[0]

tables = {}
for model in MODEL_SPECS:
    tables[model], error = load_table(model)

# -------------------------------------------------------------------
# 1. Inventory
# Expected: every model should show exists=True and a non-empty schema.
# -------------------------------------------------------------------
print("\n===== 1. TABLE INVENTORY =====")

inventory = []
for model, df in tables.items():
    if df is None:
        inventory.append((model, False, 0, "", "TABLE NOT AVAILABLE"))
    else:
        inventory.append((
            model,
            True,
            len(df.columns),
            ", ".join(df.columns),
            ""
        ))

spark.createDataFrame(
    inventory,
    ["model", "exists", "column_count", "columns", "error"]
).show(truncate=False)

# -------------------------------------------------------------------
# 2. Row counts and time coverage
# Expected:
# - min/max timestamps should match the source availability window.
# - datestamp should represent 15-minute intervals.
# -------------------------------------------------------------------
print("\n===== 2. ROW COUNTS AND TIME COVERAGE =====")

summary_rows = []

for model, df in tables.items():
    if df is None:
        continue

    if "datestamp" in df.columns:
        d = df.withColumn("_ts", F.to_timestamp(F.col("datestamp")))

        expressions = [
            F.count("*").alias("row_count"),
            F.min("_ts").alias("min_datestamp"),
            F.max("_ts").alias("max_datestamp"),
            F.approx_count_distinct("_ts").alias("approx_interval_count"),
        ]

        if "node" in df.columns:
            expressions.append(
                F.approx_count_distinct("node").alias("approx_node_count")
            )

        result = d.agg(*expressions).first().asDict()

        # Counts timestamps not aligned to 15-minute boundaries.
        misaligned = d.where(
            (F.minute("_ts") % 15 != 0) |
            (F.second("_ts") != 0)
        ).count()

        summary_rows.append((
            model,
            result.get("row_count"),
            result.get("min_datestamp"),
            result.get("max_datestamp"),
            result.get("approx_interval_count"),
            result.get("approx_node_count"),
            misaligned,
        ))

spark.createDataFrame(
    summary_rows,
    [
        "model", "row_count", "min_datestamp", "max_datestamp",
        "approx_interval_count", "approx_node_count",
        "misaligned_15m_rows",
    ],
).show(truncate=False)

# -------------------------------------------------------------------
# 3. Grain uniqueness
# Expected: duplicate_groups should be zero.
#
# Long-form tables containing metric_name/value are checked at:
# technical grain + metric_name.
# This prevents different metrics from being incorrectly treated as duplicates.
# -------------------------------------------------------------------
print("\n===== 3. GRAIN UNIQUENESS =====")

grain_rows = []

for model, df in tables.items():
    if df is None:
        continue

    configured_grain = MODEL_SPECS[model]["grain"]
    actual_grain = [c for c in configured_grain if c in df.columns]

    # Long-form metric tables require metric_name in the uniqueness grain.
    if "metric_name" in df.columns and "value" in df.columns:
        actual_grain.append("metric_name")

    missing_grain = [
        c for c in configured_grain
        if c not in df.columns
    ]

    if not actual_grain:
        grain_rows.append((
            model, "NOT CHECKED", None, None,
            ",".join(missing_grain), "No usable grain columns"
        ))
        continue

    grouped = df.groupBy(*actual_grain).count()

    duplicate_stats = grouped.where(
        F.col("count") > 1
    ).agg(
        F.count("*").alias("duplicate_groups"),
        F.max("count").alias("max_rows_per_group"),
        F.sum(F.col("count") - 1).alias("excess_rows"),
    ).first()

    grain_rows.append((
        model,
        " + ".join(actual_grain),
        duplicate_stats["duplicate_groups"] or 0,
        duplicate_stats["max_rows_per_group"] or 0,
        ",".join(missing_grain),
        "",
    ))

spark.createDataFrame(
    grain_rows,
    [
        "model", "checked_grain", "duplicate_groups",
        "max_rows_per_group", "missing_expected_keys", "notes"
    ],
).show(truncate=False)

# -------------------------------------------------------------------
# 4. Null technical keys
# Expected: zero nulls for keys required by that model's intended grain.
# -------------------------------------------------------------------
print("\n===== 4. NULL TECHNICAL KEYS =====")

null_rows = []

for model, df in tables.items():
    if df is None:
        continue

    keys = [
        c for c in MODEL_SPECS[model]["grain"]
        if c in df.columns
    ]

    if not keys:
        continue

    expressions = [
        F.sum(
            F.when(F.col(k).isNull(), 1).otherwise(0)
        ).alias(f"null_{k}")
        for k in keys
    ]

    values = df.agg(*expressions).first().asDict()
    null_rows.append(
        (model, *[values.get(f"null_{k}", 0) for k in keys])
    )

# Create one consistent output even though models have different available keys.
null_output = []
for row in null_rows:
    model = row[0]
    values = row[1:]
    null_output.append((
        model,
        sum(v or 0 for v in values),
        str(values),
    ))

spark.createDataFrame(
    null_output,
    ["model", "total_null_key_values", "null_counts_in_grain_order"]
).show(truncate=False)

# -------------------------------------------------------------------
# 5. Metric population
# Expected:
# - long-form tables: each expected metric should have rows and non-null values.
# - wide tables: expected fields should have non-null counts.
# Zero is acceptable only when the source is known to be unavailable.
# -------------------------------------------------------------------
print("\n===== 5. METRIC POPULATION =====")

population_rows = []

for model, df in tables.items():
    if df is None:
        continue

    # Long-form output: report one row per metric_name.
    if "metric_name" in df.columns and "value" in df.columns:
        metric_df = (
            df.withColumn("_value_num", F.col("value").cast("double"))
              .groupBy("metric_name")
              .agg(
                  F.count("*").alias("row_count"),
                  F.count("value").alias("non_null_value_count"),
                  F.min("_value_num").alias("min_value"),
                  F.max("_value_num").alias("max_value"),
              )
              .withColumn("model", F.lit(model))
              .select(
                  "model", "metric_name", "row_count",
                  "non_null_value_count", "min_value", "max_value"
              )
        )
        population_rows.append(metric_df)

    # Wide output: report only expected fields that exist.
    else:
        present = [
            c for c in MODEL_SPECS[model]["measures"]
            if c in df.columns
        ]

        for measure in present:
            result = df.agg(
                F.count("*").alias("row_count"),
                F.count(F.col(measure)).alias("non_null_count"),
                F.min(F.col(measure).cast("double")).alias("min_value"),
                F.max(F.col(measure).cast("double")).alias("max_value"),
            ).first()

            population_rows.append(
                spark.createDataFrame([(
                    model,
                    measure,
                    result["row_count"],
                    result["non_null_count"],
                    result["min_value"],
                    result["max_value"],
                )], T.StructType([
                    T.StructField("model", T.StringType(), True),
                    T.StructField("metric_name", T.StringType(), True),
                    T.StructField("row_count", T.LongType(), True),
                    T.StructField("non_null_value_count", T.LongType(), True),
                    T.StructField("min_value", T.DoubleType(), True),
                    T.StructField("max_value", T.DoubleType(), True),
                ]))
            )

if population_rows:
    population = population_rows[0]
    for p in population_rows[1:]:
        population = population.unionByName(p)
    population.orderBy("model", "metric_name").show(
        500, truncate=False
    )
else:
    print("No metric columns were found.")

# -------------------------------------------------------------------
# 6. Negative-value checks
# Expected:
# - utilization, traffic, demand, and derived deltas should normally be >= 0.
# - negative cumulative-counter transitions may indicate resets.
# - investigate nonzero results; do not automatically classify resets as defects.
# -------------------------------------------------------------------
print("\n===== 6. NEGATIVE VALUE CHECKS =====")

negative_rows = []

for model, df in tables.items():
    if df is None:
        continue

    if "metric_name" in df.columns and "value" in df.columns:
        result = (
            df.withColumn("_value_num", F.col("value").cast("double"))
              .where(F.col("_value_num") < 0)
              .groupBy("metric_name")
              .count()
        )

        for r in result.collect():
            negative_rows.append((
                model,
                r["metric_name"],
                r["count"],
            ))
    else:
        present = [
            c for c in MODEL_SPECS[model]["measures"]
            if c in df.columns
        ]

        for measure in present:
            count_negative = df.where(
                F.col(measure).cast("double") < 0
            ).count()

            negative_rows.append((model, measure, count_negative))

if negative_rows:
    spark.createDataFrame(
        negative_rows,
        ["model", "metric_or_column", "negative_rows"]
    ).show(500, truncate=False)
else:
    print("No negative values found.")

# -------------------------------------------------------------------
# 7. Network ratio checks
# Expected:
# - ratios should be nonnegative.
# - NULL is acceptable when ingress is zero or source data is unavailable.
# - these ratios should not be summed across node-level classifier rows.
# -------------------------------------------------------------------
print("\n===== 7. NETWORK RATIO CHECKS =====")

ratio_rows = []

for model, df in tables.items():
    if df is None:
        continue

    ratio_columns = [
        c for c in df.columns
        if "ratio" in c.lower()
    ]

    for ratio in ratio_columns:
        result = df.agg(
            F.count("*").alias("row_count"),
            F.count(F.col(ratio)).alias("non_null_count"),
            F.sum(
                F.when(F.col(ratio).cast("double") < 0, 1)
                 .otherwise(0)
            ).alias("negative_count"),
            F.min(F.col(ratio).cast("double")).alias("min_ratio"),
            F.max(F.col(ratio).cast("double")).alias("max_ratio"),
        ).first()

        ratio_rows.append((
            model,
            ratio,
            result["row_count"],
            result["non_null_count"],
            result["negative_count"],
            result["min_ratio"],
            result["max_ratio"],
        ))

if ratio_rows:
    spark.createDataFrame(
        ratio_rows,
        T.StructType([
            T.StructField("model", T.StringType(), True),
            T.StructField("ratio_column", T.StringType(), True),
            T.StructField("row_count", T.LongType(), True),
            T.StructField("non_null_count", T.LongType(), True),
            T.StructField("negative_count", T.LongType(), True),
            T.StructField("min_ratio", T.DoubleType(), True),
            T.StructField("max_ratio", T.DoubleType(), True),
        ]),
    ).show(truncate=False)
else:
    print("No ratio columns found.")

# -------------------------------------------------------------------
# 8. Representative rows
# Expected:
# - review a few recent rows for correct keys, values, and source freshness.
# - classifier tables should retain node grain.
# -------------------------------------------------------------------
print("\n===== 8. RECENT SAMPLE ROWS =====")

for model, df in tables.items():
    if df is not None and "datestamp" in df.columns:
        print(f"\n--- {model} ---")
        df.orderBy(F.col("datestamp").desc()).limit(5).show(
            truncate=False
        )
```
