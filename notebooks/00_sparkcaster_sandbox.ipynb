```python
#install needed packages

import importlib
import subprocess
import sys

def is_module_available(module_name: str) -> bool:
    """Return True if the importable module exists; handles dotted names safely."""
    try:
        return importlib.util.find_spec(module_name) is not None
    except ModuleNotFoundError:
        return False

def install_if_missing(pip_name: str, import_name: str = None):
    """
    Install `pip_name` only if `import_name` (or `pip_name` if None) is not importable.
    Use this to handle cases where pip/distribution names differ from import names.
    """
    import_name = import_name or pip_name
    if not is_module_available(import_name):
        print(f"Installing {pip_name} ...")
        subprocess.check_call([sys.executable, "-m", "pip", "install", pip_name])
    else:
        print(f"{pip_name} already installed.")

# --- Packages for Nessie/Iceberg access ---
# simple 1:1 cases
install_if_missing("keyring", "keyring")
install_if_missing("ipython-secrets", "ipython_secrets")
install_if_missing("pyarrow", "pyarrow")
install_if_missing("fsspec", "fsspec")
install_if_missing("s3fs", "s3fs")
install_if_missing("statsmodels", "statsmodels")
install_if_missing("matplotlib", "matplotlib")
install_if_missing("scikit-learn", "sklearn")
install_if_missing("xgboost", "xgboost")

# special cases
# keyrings.cryptfile installs a plugin under the 'keyrings' package; checking the root avoids ModuleNotFoundError
install_if_missing("keyrings.cryptfile", "keyrings")

# Core viz + scaling pins for Python 3.11.11
install_if_missing("bokeh==3.6.2", "bokeh")
install_if_missing("jupyter_bokeh", "jupyter_bokeh")
install_if_missing("panel==1.5.2", "panel")
install_if_missing("holoviews==1.19.0", "holoviews")
install_if_missing("hvplot==0.10.0", "hvplot")
install_if_missing("datashader==0.16.3", "datashader")
install_if_missing("dask[dataframe]==2024.9.1", "dask")
install_if_missing("distributed==2024.9.1", "distributed")
install_if_missing("reportlab", "reportlab")


print("\n✓ All packages installed/verified")


```

```python
spark.sql('''
WITH base AS (
    SELECT
        DATE_TRUNC('month', datestamp) AS month_start,
        DATE_TRUNC('day',   datestamp) AS cal_day,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        COALESCE(CAST(is_sunk_running AS INT), -1) AS is_sunk_running,
        node,
        LEAST(tensor_util, 1.0) AS tensor_util_capped
    FROM sandbox.sandbox_finance.dcgm_metrics_raw_imputed_v2
    WHERE datestamp >= '2026-04-01'
      AND datestamp <  '2026-07-01'
      -- AND is_hpc_verification_job = 0
),

node_daily_with_sunk AS (
    SELECT
        month_start,
        cal_day,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        is_sunk_running,
        node,
        AVG(tensor_util_capped) AS tensor_util_node
    FROM base
    GROUP BY
        month_start,
        cal_day,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        is_sunk_running,
        node
),

node_daily_without_sunk AS (
    SELECT
        month_start,
        cal_day,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        node,
        AVG(tensor_util_capped) AS tensor_util_node
    FROM base
    GROUP BY
        month_start,
        cal_day,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        node
),

monthly_with_sunk AS (
    SELECT
        month_start,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        is_sunk_running,
        element_at(percentile_approx(tensor_util_node, array(0.50, 0.95), 1000), 1) AS tensor_util_p50,
        element_at(percentile_approx(tensor_util_node, array(0.50, 0.95), 1000), 2) AS tensor_util_p95,
        COUNT(*) AS node_day_rows
    FROM node_daily_with_sunk
    GROUP BY
        month_start,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        is_sunk_running
),

monthly_without_sunk AS (
    SELECT
        month_start,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        element_at(percentile_approx(tensor_util_node, array(0.50, 0.95), 1000), 1) AS tensor_util_p50,
        element_at(percentile_approx(tensor_util_node, array(0.50, 0.95), 1000), 2) AS tensor_util_p95,
        COUNT(*) AS node_day_rows
    FROM node_daily_without_sunk
    GROUP BY
        month_start,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode
),

split_pivot AS (
    SELECT
        month_start,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode,
        AVG(tensor_util_p50) AS avg_split_tensor_util_p50,
        AVG(tensor_util_p95) AS avg_split_tensor_util_p95,
        MAX(CASE WHEN is_sunk_running = 0  THEN tensor_util_p50 END) AS sunk0_tensor_util_p50,
        MAX(CASE WHEN is_sunk_running = 1  THEN tensor_util_p50 END) AS sunk1_tensor_util_p50,
        MAX(CASE WHEN is_sunk_running = -1 THEN tensor_util_p50 END) AS sunknull_tensor_util_p50,
        MAX(CASE WHEN is_sunk_running = 0  THEN tensor_util_p95 END) AS sunk0_tensor_util_p95,
        MAX(CASE WHEN is_sunk_running = 1  THEN tensor_util_p95 END) AS sunk1_tensor_util_p95,
        MAX(CASE WHEN is_sunk_running = -1 THEN tensor_util_p95 END) AS sunknull_tensor_util_p95,
        COUNT(*) AS split_group_count,
        SUM(node_day_rows) AS split_node_day_rows
    FROM monthly_with_sunk
    GROUP BY
        month_start,
        region,
        customer_name,
        customer_segment,
        product_resolved,
        product_segment,
        is_training,
        is_multinode
)

SELECT
    w.month_start,
    w.region,
    w.customer_name,
    w.customer_segment,
    w.product_resolved,
    w.product_segment,
    w.is_training,
    w.is_multinode,

    w.tensor_util_p50 AS pooled_tensor_util_p50,
    s.avg_split_tensor_util_p50,
    s.sunk0_tensor_util_p50,
    s.sunk1_tensor_util_p50,
    s.sunknull_tensor_util_p50,
    s.avg_split_tensor_util_p50 - w.tensor_util_p50 AS delta_p50,

    w.tensor_util_p95 AS pooled_tensor_util_p95,
    s.avg_split_tensor_util_p95,
    s.sunk0_tensor_util_p95,
    s.sunk1_tensor_util_p95,
    s.sunknull_tensor_util_p95,
    s.avg_split_tensor_util_p95 - w.tensor_util_p95 AS delta_p95,

    w.node_day_rows AS pooled_node_day_rows,
    s.split_node_day_rows,
    s.split_group_count
FROM monthly_without_sunk w
LEFT JOIN split_pivot s
  ON  w.month_start       <=> s.month_start
  AND w.region            <=> s.region
  AND w.customer_name     <=> s.customer_name
  AND w.customer_segment  <=> s.customer_segment
  AND w.product_resolved  <=> s.product_resolved
  AND w.product_segment   <=> s.product_segment
  AND w.is_training       <=> s.is_training
  AND w.is_multinode      <=> s.is_multinode
ORDER BY
    ABS(s.avg_split_tensor_util_p95 - w.tensor_util_p95) DESC,
    ABS(s.avg_split_tensor_util_p50 - w.tensor_util_p50) DESC
LIMIT 200;
''').show()
```

```python
spark.sql('''
  SELECT
    SUM(CASE WHEN is_training       IS NULL THEN 1 ELSE 0 END) AS null_is_training,
    SUM(CASE WHEN is_sunk_running   IS NULL THEN 1 ELSE 0 END) AS null_is_sunk,
    SUM(CASE WHEN is_multinode      IS NULL THEN 1 ELSE 0 END) AS null_is_multinode,
    SUM(CASE WHEN flag_is_coreweave IS NULL THEN 1 ELSE 0 END) AS null_flag_cw,
    SUM(CASE WHEN region_rollup     IS NULL THEN 1 ELSE 0 END) AS null_region_rollup,
    SUM(CASE WHEN product_segment   IS NULL THEN 1 ELSE 0 END) AS null_prod_segment,
    COUNT(*) AS n
  FROM sandbox.sandbox_finance.dcgm_metrics_summary_imputed_daily_v2
  WHERE day < DATE '2025-05-01'
''').show(truncate=False)
```
